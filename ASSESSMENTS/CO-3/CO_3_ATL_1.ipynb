```python
import re
from collections import Counter

# ---------------- CORPUS ----------------

corpus = [
    "The student is intelligent",
    "The student is hardworking",
    "The student is learning",
    "The student reads books",
    "The student writes programs",
    "The teacher is intelligent",
    "The teacher is teaching",
    "The teacher reads books",
    "The teacher explains lessons",
    "The programmer writes programs",
    "The programmer is learning",
    "The programmer writes code",
    "The computer is fast",
    "The computer is powerful",
    "The computer runs programs",
    "The student learns programming",
    "The student writes code",
    "Students are learning programming",
    "Students are reading books"
]

# Convert to words
sentences = [
    re.findall(r"[a-z]+", s.lower())
    for s in corpus
]

# ---------------- COUNT N-GRAMS ----------------

unigram = Counter()
bigram = Counter()
trigram = Counter()

for words in sentences:

    for word in words:
        unigram[word] += 1

    for i in range(len(words) - 1):
        bigram[(words[i], words[i + 1])] += 1

    for i in range(len(words) - 2):
        trigram[
            (words[i], words[i + 1], words[i + 2])
        ] += 1


# ---------------- PROBABILITIES ----------------

def unigram_prob(word):
    return unigram[word] / sum(unigram.values())


def bigram_prob(w1, w2):

    count = bigram[(w1, w2)]

    total = sum(
        c for (a, b), c in bigram.items()
        if a == w1
    )

    if total == 0:
        return 0

    return count / total


def trigram_prob(w1, w2, w3):

    count = trigram[(w1, w2, w3)]

    total = sum(
        c for (a, b, c), c in trigram.items()
        if a == w1 and b == w2
    )

    if total == 0:
        return 0

    return count / total


# ---------------- BACKOFF ----------------

def backoff(w1, w2, word):

    p = trigram_prob(w1, w2, word)

    if p > 0:
        return p, "Trigram"

    p = bigram_prob(w2, word)

    if p > 0:
        return p, "Bigram"

    p = unigram_prob(word)

    if p > 0:
        return p, "Unigram"

    return 0, "Unseen"


# ---------------- INTERPOLATION ----------------

def interpolation(w1, w2, word):

    p1 = unigram_prob(word)
    p2 = bigram_prob(w2, word)
    p3 = trigram_prob(w1, w2, word)

    # Weights
    probability = (
        0.2 * p1 +
        0.3 * p2 +
        0.5 * p3
    )

    return probability


# ---------------- PREDICTION ----------------

def predict(sentence):

    words = re.findall(
        r"[a-z]+",
        sentence.lower()
    )

    if len(words) < 2:
        print("Enter at least two words.")
        return

    w1 = words[-2]
    w2 = words[-1]

    print("\nContext:", w1, w2)

    # Unsmoothed
    print("\n1. Unsmoothed Trigram")

    result = []

    for word in unigram:

        p = trigram_prob(w1, w2, word)

        if p > 0:
            result.append((word, p))

    result.sort(
        key=lambda x: x[1],
        reverse=True
    )

    for word, p in result[:5]:
        print(word, round(p, 4))


    # Backoff
    print("\n2. Backoff")

    result = []

    for word in unigram:

        p, method = backoff(
            w1, w2, word
        )

        if p > 0:
            result.append(
                (word, p, method)
            )

    result.sort(
        key=lambda x: x[1],
        reverse=True
    )

    for word, p, method in result[:5]:
        print(
            word,
            round(p, 4),
            method
        )


    # Interpolation
    print("\n3. Deleted Interpolation")

    result = []

    for word in unigram:

        p = interpolation(
            w1, w2, word
        )

        if p > 0:
            result.append(
                (word, p)
            )

    result.sort(
        key=lambda x: x[1],
        reverse=True
    )

    for word, p in result[:5]:
        print(
            word,
            round(p, 4)
        )


# ---------------- ZERO PROBABILITY ----------------

def zero_test(w1, w2, word):

    print("\nProbability Comparison")

    p3 = trigram_prob(
        w1, w2, word
    )

    p2 = bigram_prob(
        w2, word
    )

    p1 = unigram_prob(
        word
    )

    print("Trigram :", round(p3, 4))
    print("Bigram  :", round(p2, 4))
    print("Unigram :", round(p1, 4))

    p, method = backoff(
        w1, w2, word
    )

    print(
        "Backoff :",
        round(p, 4),
        "(" + method + ")"
    )

    p = interpolation(
        w1, w2, word
    )

    print(
        "Interpolation :",
        round(p, 4)
    )


# ---------------- MENU ----------------

while True:

    print("\n==============================")
    print(" N-GRAM LANGUAGE MODEL")
    print("==============================")
    print("1. Predict next word")
    print("2. Test zero probability")
    print("3. Exit")

    choice = input("Enter choice: ")

    if choice == "1":

        sentence = input(
            "Enter sentence: "
        )

        predict(sentence)

    elif choice == "2":

        w1 = input("First word: ").lower()
        w2 = input("Second word: ").lower()
        word = input("Next word: ").lower()

        zero_test(
            w1,
            w2,
            word
        )

    elif choice == "3":

        print("Program terminated.")
        break

    else:

        print("Invalid choice.")
```

```python
import re
import math
from collections import Counter

# Training data
train = [
    "The student is intelligent",
    "The student is hardworking",
    "The student is learning",
    "The student reads books",
    "The student writes programs",
    "The teacher is intelligent",
    "The teacher is teaching",
    "The teacher reads books",
    "The teacher explains lessons",
    "The programmer writes programs",
    "The programmer is learning",
    "The programmer writes code",
    "The computer is fast",
    "The computer is powerful",
    "The computer runs programs",
    "The student learns programming"
]

# Test data
test = [
    "The student is intelligent",
    "The teacher is teaching",
    "The programmer writes programs",
    "The computer is fast"
]

# Tokenize
def words(text):
    return re.findall(r"[a-z]+", text.lower())

# Count n-grams
unigram = Counter()
bigram = Counter()
trigram = Counter()

for sentence in train:
    w = words(sentence)

    for word in w:
        unigram[word] += 1

    for i in range(len(w)-1):
        bigram[(w[i], w[i+1])] += 1

    for i in range(len(w)-2):
        trigram[(w[i], w[i+1], w[i+2])] += 1


# Unigram probability
def uni_prob(word):
    return unigram[word] / sum(unigram.values())


# Bigram probability
def bi_prob(w1, w2):
    count = bigram[(w1, w2)]

    total = sum(
        c for (a, b), c in bigram.items()
        if a == w1
    )

    if total == 0:
        return 0

    return count / total


# Trigram probability
def tri_prob(w1, w2, w3):
    count = trigram[(w1, w2, w3)]

    total = sum(
        c for (a, b, c), c in trigram.items()
        if a == w1 and b == w2
    )

    if total == 0:
        return 0

    return count / total


# Calculate entropy
def entropy(n):

    values = []

    for sentence in test:
        w = words(sentence)

        if n == 1:
            for word in w:
                p = uni_prob(word)

                if p > 0:
                    values.append(-math.log2(p))

        elif n == 2:
            for i in range(len(w)-1):
                p = bi_prob(w[i], w[i+1])

                if p > 0:
                    values.append(-math.log2(p))

        elif n == 3:
            for i in range(len(w)-2):
                p = tri_prob(
                    w[i],
                    w[i+1],
                    w[i+2]
                )

                if p > 0:
                    values.append(-math.log2(p))

    return sum(values) / len(values)


# Smoothed unigram entropy
def smooth_entropy():

    total = sum(unigram.values())
    vocabulary = len(unigram)

    values = []

    for sentence in test:
        for word in words(sentence):

            p = (
                unigram[word] + 1
            ) / (
                total + vocabulary
            )

            values.append(-math.log2(p))

    return sum(values) / len(values)


# Next word prediction
def predict(sentence):

    w = words(sentence)

    if len(w) < 2:
        print("Enter at least 2 words")
        return

    w1 = w[-2]
    w2 = w[-1]

    result = []

    for word in unigram:

        p = tri_prob(w1, w2, word)

        if p > 0:
            result.append((word, p))

    result.sort(
        key=lambda x: x[1],
        reverse=True
    )

    print("\nNext Word Prediction:")

    for word, p in result[:5]:
        print(word, round(p, 4))


# ---------------- MAIN ----------------

print("=" * 50)
print("N-GRAM ENTROPY")
print("=" * 50)

u = entropy(1)
b = entropy(2)
t = entropy(3)
s = smooth_entropy()

print("\nEntropy Results")

print("Unigram :", round(u, 4))
print("Bigram  :", round(b, 4))
print("Trigram :", round(t, 4))
print("Smoothed:", round(s, 4))

# Prediction
print("\nPrediction")

sentence = input(
    "Enter sentence: "
)

predict(sentence)

# Interpretation
print("\nInterpretation")

if t < b:
    print("Trigram has lower entropy.")
else:
    print("Bigram has lower entropy.")

print("Lower entropy = more predictable text.")
print("Smoothing prevents zero probability.")
```

```python
import re
from collections import Counter

# Training data
training = [
    ("the", "DT"),
    ("student", "NN"),
    ("is", "VBZ"),
    ("intelligent", "JJ"),
    ("reads", "VBZ"),
    ("books", "NNS"),
    ("teacher", "NN"),
    ("teaches", "VBZ"),
    ("students", "NNS"),
    ("programmer", "NN"),
    ("writes", "VBZ"),
    ("code", "NN"),
    ("computer", "NN"),
    ("fast", "JJ"),
    ("learns", "VBZ"),
    ("programming", "NN")
]

# Store known words and their tags
word_tags = {}

for word, tag in training:
    word_tags[word] = tag


# ---------------- RULE BASED ----------------

def rule_based(words):

    result = []

    for word in words:

        w = word.lower()

        if w in ["the", "a", "an"]:
            tag = "DT"

        elif w in ["he", "she", "it", "they", "we", "i"]:
            tag = "PRP"

        elif w in ["is", "am", "are", "was", "were"]:
            tag = "VBZ"

        elif w in ["in", "on", "at", "by", "with"]:
            tag = "IN"

        elif w.endswith("ing"):
            tag = "VBG"

        elif w.endswith("ed"):
            tag = "VBD"

        elif w.endswith("ly"):
            tag = "RB"

        elif w.endswith("s"):
            tag = "NNS"

        elif w in ["intelligent", "fast", "good", "beautiful"]:
            tag = "JJ"

        elif w in word_tags:
            tag = word_tags[w]

        else:
            tag = "NN"

        result.append((word, tag))

    return result


# ---------------- STOCHASTIC ----------------

def stochastic(words):

    result = []

    for word in words:

        w = word.lower()

        # Use training tag if word is known
        if w in word_tags:
            tag = word_tags[w]

        # Otherwise use simple probability/rules
        elif w.endswith("ing"):
            tag = "VBG"

        elif w.endswith("ed"):
            tag = "VBD"

        elif w.endswith("s"):
            tag = "NNS"

        else:
            tag = "NN"

        result.append((word, tag))

    return result


# ---------------- TRANSFORMATION ----------------

def transformation(words):

    result = rule_based(words)

    for i in range(1, len(result)):

        word, tag = result[i]

        previous = words[i - 1].lower()

        # After "is/am/are" + ing = VBG
        if previous in ["is", "am", "are"]:
            if word.lower().endswith("ing"):
                result[i] = (word, "VBG")

        # After pronoun, word can be verb
        if previous in ["he", "she", "they", "we", "i"]:
            if word.lower().endswith("s"):
                result[i] = (word, "VBZ")

    return result


# ---------------- DISPLAY ----------------

def show(title, result):

    print("\n" + title)
    print("-" * 30)

    for word, tag in result:
        print(word, "->", tag)


# ---------------- MAIN ----------------

print("=" * 40)
print("POS TAGGING SYSTEM")
print("=" * 40)

sentence = input("Enter a sentence: ")

words = re.findall(
    r"[A-Za-z]+",
    sentence
)

rule = rule_based(words)
stochastic_result = stochastic(words)
transform = transformation(words)

show(
    "RULE-BASED TAGGING",
    rule
)

show(
    "STOCHASTIC TAGGING",
    stochastic_result
)

show(
    "TRANSFORMATION-BASED TAGGING",
    transform
)

# Tag information
print("\nTAG INFORMATION")
print("----------------")
print("NN  = Noun")
print("NNS = Plural Noun")
print("VBZ = Verb")
print("VBG = Verb + ing")
print("VBD = Past Verb")
print("JJ  = Adjective")
print("RB  = Adverb")
print("PRP = Pronoun")
print("DT  = Determiner")
print("IN  = Preposition")
```
