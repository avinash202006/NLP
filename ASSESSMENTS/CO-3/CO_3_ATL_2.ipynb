```python
import math
from collections import Counter

corpus = "Data science is powerful, data science drives innovation, data science is evolving."

words = corpus.lower().replace(",", "").replace(".", "").split()

unigram = Counter(words)
bigram = Counter(zip(words, words[1:]))
trigram = Counter(zip(words, words[1:], words[2:]))

# Q1: MLE P(science | data)
p1 = bigram[("data", "science")] / unigram["data"]
print("1. P(science | data) =", p1)

# Q2: Backoff P(improves | data science)
word = "improves"

if trigram[("data", "science", word)] > 0:
    probability = trigram[("data", "science", word)] / bigram[("data", "science")]
elif bigram[("science", word)] > 0:
    probability = bigram[("science", word)] / unigram["science"]
elif unigram[word] > 0:
    probability = unigram[word] / len(words)
else:
    probability = 0

print("2. Backoff P(improves | data science) =", probability)

# Q3: Deleted Interpolation
p_trigram = trigram[("data", "science", "is")] / bigram[("data", "science")]
p_bigram = bigram[("science", "is")] / unigram["science"]
p_unigram = unigram["is"] / len(words)

result = (0.5 * p_trigram) + (0.3 * p_bigram) + (0.2 * p_unigram)

print("3. Interpolation Probability =", result)

# Q4: Entropy
p_is = 0.66
p_drives = 0.33

entropy = -(p_is * math.log2(p_is) +
            p_drives * math.log2(p_drives))

print("4. Entropy =", entropy, "bits")
```

```python
sentences = [
    "Book a flight ticket now.",
    "This book is interesting."
]

# POS tags
tags = {
    "Book": "VB",
    "a": "DT",
    "flight": "NN",
    "ticket": "NN",
    "now": "RB",
    "This": "DT",
    "book": "NN",
    "is": "VBZ",
    "interesting": "JJ"
}

# Display POS tags
for sentence in sentences:
    print("\nSentence:", sentence)
    for word in sentence.replace(".", "").split():
        print(word, "->", tags[word])

# HMM probabilities
p_book_vb = 0.6
p_book_nn = 0.4

p_start_vb = 0.5
p_start_nn = 0.5

# Calculate probabilities
prob_vb = p_start_vb * p_book_vb
prob_nn = p_start_nn * p_book_nn

print("\nHMM Calculation")
print("P(Book, VB) =", prob_vb)
print("P(Book, NN) =", prob_nn)

# Prediction
if prob_vb > prob_nn:
    print("Book is predicted as: VB")
else:
    print("Book is predicted as: NN")

# Normalized probabilities
total = prob_vb + prob_nn

print("\nNormalized Probability")
print("P(VB | Book) =", prob_vb / total)
print("P(NN | Book) =", prob_nn / total)
```

```python
import math

# Sentence and initial tags
sentence = ["economic", "growth", "increases", "employment"]
initial_tags = ["JJ", "NN", "NNS", "NN"]

print("INITIAL POS TAGS")
for word, tag in zip(sentence, initial_tags):
    print(word, "->", tag)

# Apply transformation rule
corrected_tags = initial_tags.copy()

for i in range(1, len(corrected_tags)):
    if corrected_tags[i] == "NNS" and corrected_tags[i - 1] == "NN":
        corrected_tags[i] = "VBZ"

print("\nCORRECTED POS TAGS")
for word, tag in zip(sentence, corrected_tags):
    print(word, "->", tag)

# Word frequencies
frequency = {
    "economic": 120,
    "growth": 450,
    "increases": 210,
    "employment": 380
}

total = sum(frequency.values())

print("\nWORD FREQUENCY")
for word, count in frequency.items():
    probability = count / total
    print(word, ":", count, "Probability =", round(probability, 4))

print("Total =", total)

# Entropy of word distribution
entropy = 0

for count in frequency.values():
    p = count / total
    entropy -= p * math.log2(p)

print("\nWORD DISTRIBUTION ENTROPY =", round(entropy, 4), "bits")

# Tag uncertainty
before = [0.5, 0.5]
after = [0.9, 0.1]

entropy_before = -sum(p * math.log2(p) for p in before)
entropy_after = -sum(p * math.log2(p) for p in after)

print("\nTAG UNCERTAINTY")
print("Before transformation =", round(entropy_before, 4), "bits")
print("After transformation =", round(entropy_after, 4), "bits")

if entropy_after < entropy_before:
    print("Uncertainty decreased after transformation.")
```
