1)Biomedical Dataset: Porter Stemming Error Analysis

```python
from nltk.stem import PorterStemmer
import re

stemmer = PorterStemmer()

text = """
Infection is a common clinical condition.
The infectious disease was treated successfully.
The patient was infected with a bacterial infection.
Researchers study infection and infectious diseases.
The patient recovered from the infection.
Doctors identified an infectious agent.
The infected patient received treatment.
The study investigates new methods to prevent infection.
"""

# -------------------------------
# 1. PREPROCESSING
# -------------------------------

text = text.lower()
text = re.sub(r'[^a-z\s]', '', text)

words = text.split()

print("PREPROCESSED WORDS")
print("------------------")
print(words)


# -------------------------------
# 2. APPLY PORTER STEMMER
# -------------------------------

print("\nWORD AND PORTER STEM")
print("--------------------")

for word in words:
    stem = stemmer.stem(word)
    print(f"{word:15} -> {stem}")


# -------------------------------
# 3. ANALYZE IMPORTANT WORDS
# -------------------------------

test_words = [
    "infect",
    "infected",
    "infection",
    "infectious"
]

print("\nMORPHOLOGICAL ERROR ANALYSIS")
print("----------------------------")

for word in test_words:

    stem = stemmer.stem(word)

    if word == "infect":
        observation = "Root word"

    elif word == "infected":
        observation = "Inflectional suffix '-ed' removed"

    elif word == "infection":
        observation = "Derivational suffix '-ion' removed"

    elif word == "infectious":
        observation = "Derivational information lost"

    print(
        f"{word:12} -> {stem:10} : {observation}"
    )


# -------------------------------
# 4. IDENTIFY STEMMING ERRORS
# -------------------------------

print("\nSTEMMING ERRORS")
print("--------------")

for word in test_words:

    stem = stemmer.stem(word)

    if word == "infectious" and stem != "infect":
        print(
            f"{word} -> {stem} : Incorrect/unnatural stem"
        )

    elif word == "infection":
        print(
            f"{word} -> {stem} : Derivational information removed"
        )

    elif word == "infected":
        print(
            f"{word} -> {stem} : Inflectional information removed"
        )

    else:
        print(
            f"{word} -> {stem} : No major stemming error"
        )


# -------------------------------
# 5. INFLECTIONAL VS DERIVATIONAL
# -------------------------------

print("\nINFLECTIONAL VS DERIVATIONAL MORPHOLOGY")
print("----------------------------------------")

print("Inflectional:")
print("infect -> infected")
print("The suffix '-ed' indicates grammatical tense.")
print("Removing it can help retrieve related word forms.")

print("\nDerivational:")
print("infect -> infection")
print("infect -> infectious")
print("These forms create different words and grammatical categories.")
print("Blind stemming can remove important biomedical meaning.")


# -------------------------------
# 6. FINAL CONCLUSION
# -------------------------------

print("\nCONCLUSION")
print("----------")

print("Porter Stemmer is useful for simple word normalization.")
print("However, it does not understand the actual meaning of words.")
print("For biomedical search, blindly stemming all words can")
print("remove important derivational information.")
print("Lemmatization or domain-specific morphological analysis")
print("is therefore a better approach.")
```

2)Finite-State Morphological Parser

```python
import re

# Social media sample data
text = """
The happiest customer gave an unbelievable review.
Running customers liked the reordering option.
Smartphones are useful.
The unreadable message was deleted.
"""

# Words to test
words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

test_words = [
    "happiest",
    "unbelievable",
    "running",
    "reordering",
    "smartphones",
    "unreadable"
]


# Original simple parser
def old_parser(word):

    parts = []

    if word.startswith("un"):
        parts.append("un")
        word = word[2:]

    elif word.startswith("re"):
        parts.append("re")
        word = word[2:]

    if word.endswith("ing"):
        parts.append("ing")
        word = word[:-3]

    elif word.endswith("est"):
        parts.append("est")
        word = word[:-3]

    elif word.endswith("able"):
        parts.append("able")
        word = word[:-4]

    elif word.endswith("s"):
        parts.append("s")
        word = word[:-1]

    return word, parts


# Improved parser
def new_parser(word):

    prefixes = []
    suffixes = []

    # Prefix transitions
    if word.startswith("un"):
        prefixes.append("un")
        word = word[2:]

    if word.startswith("re"):
        prefixes.append("re")
        word = word[2:]

    # Suffix transitions
    if word.endswith("iest"):
        suffixes.append("iest")
        word = word[:-4]

    elif word.endswith("ing"):
        suffixes.append("ing")
        word = word[:-3]

    elif word.endswith("able"):
        suffixes.append("able")
        word = word[:-4]

    elif word.endswith("s"):
        suffixes.append("s")
        word = word[:-1]

    return word, prefixes, suffixes


print("FINITE-STATE MORPHOLOGICAL PARSER")
print("----------------------------------")

print("\nOriginal Parser:")

for word in test_words:

    root, affixes = old_parser(word)

    print(
        word,
        "-> Root:",
        root,
        "Affixes:",
        affixes
    )

print("\nImproved Parser:")

for word in test_words:

    root, prefixes, suffixes = new_parser(word)

    print(
        word,
        "-> Root:",
        root,
        "Prefixes:",
        prefixes,
        "Suffixes:",
        suffixes
    )
```

3)News Classification Stemming

```python
import time
import re

from nltk.stem import PorterStemmer
from nltk.stem import WordNetLemmatizer
import nltk

from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Download WordNet
nltk.download("wordnet")
nltk.download("omw-1.4")


# Load two news categories
categories = [
    "comp.sys.ibm.pc.hardware",
    "rec.sport.baseball"
]

train = fetch_20newsgroups(
    subset="train",
    categories=categories,
    remove=("headers", "footers", "quotes")
)

test = fetch_20newsgroups(
    subset="test",
    categories=categories,
    remove=("headers", "footers", "quotes")
)


stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()


# No stemming
def normal_text(text):

    return text.lower()


# Porter stemming
def stem_text(text):

    words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

    return " ".join(
        stemmer.stem(word)
        for word in words
    )


# Lemmatization
def lemma_text(text):

    words = re.findall(r'\b[a-zA-Z]+\b', text.lower())

    return " ".join(
        lemmatizer.lemmatize(word)
        for word in words
    )


def classify(train_text, test_text, name):

    start = time.time()

    vectorizer = TfidfVectorizer(
        max_features=10000
    )

    X_train = vectorizer.fit_transform(train_text)
    X_test = vectorizer.transform(test_text)

    model = LogisticRegression(
        max_iter=1000
    )

    model.fit(X_train, train.target)

    prediction = model.predict(X_test)

    accuracy = accuracy_score(
        test.target,
        prediction
    )

    vocabulary = len(
        vectorizer.get_feature_names_out()
    )

    elapsed = time.time() - start

    print("\n", name)
    print("----------------------")
    print("Vocabulary Size:", vocabulary)
    print("Accuracy:", round(accuracy * 100, 2), "%")
    print("Processing Time:", round(elapsed, 2), "seconds")


print("NEWS CLASSIFICATION ERROR ANALYSIS")
print("----------------------------------")

# Original
train_original = [
    normal_text(x) for x in train.data
]

test_original = [
    normal_text(x) for x in test.data
]

classify(
    train_original,
    test_original,
    "WITHOUT STEMMING"
)


# Porter stemming
train_stem = [
    stem_text(x) for x in train.data
]

test_stem = [
    stem_text(x) for x in test.data
]

classify(
    train_stem,
    test_stem,
    "PORTER STEMMING"
)


# Lemmatization
train_lemma = [
    lemma_text(x) for x in train.data
]

test_lemma = [
    lemma_text(x) for x in test.data
]

classify(
    train_lemma,
    test_lemma,
    "LEMMATIZATION"
)


# Individual word analysis
print("\nPORTER STEMMING ERROR ANALYSIS")
print("------------------------------")

words = [
    "organization",
    "organizer",
    "organizing",
    "organized",
    "organization's"
]

for word in words:

    clean = re.sub(
        r'[^a-zA-Z]',
        '',
        word
    )

    print(
        word,
        "->",
        stemmer.stem(clean)
    )
```

Error Analysis of a Finite-State Morphological Parser

```python
# Question 6: Finite-State Morphological Parser

import nltk

# Download WordNet lexical database
nltk.download('wordnet')
nltk.download('omw-1.4')

from nltk.corpus import wordnet as wn


# -----------------------------------------
# Morphological Parser
# -----------------------------------------

def parser(word):
    word = word.lower()

    # Irregular plurals
    irregular = {
        "children": ("child", "Plural Noun"),
        "men": ("man", "Plural Noun"),
        "women": ("woman", "Plural Noun"),
        "mice": ("mouse", "Plural Noun"),
        "geese": ("goose", "Plural Noun"),
        "feet": ("foot", "Plural Noun"),
        "teeth": ("tooth", "Plural Noun"),
        "people": ("person", "Plural Noun"),
        "oxen": ("ox", "Plural Noun")
    }

    if word in irregular:
        return irregular[word]

    # Words ending in -ies
    # Example: cities -> city
    if word.endswith("ies") and len(word) > 3:
        return word[:-3] + "y", "Plural Noun"

    # Words ending in -es
    # Examples: boxes -> box
    #           buses -> bus
    if word.endswith("es") and len(word) > 2:
        base = word[:-2]

        # Check common -es plural patterns
        if (word.endswith("ses") or
            word.endswith("xes") or
            word.endswith("zes") or
            word.endswith("ches") or
            word.endswith("shes")):
            return base, "Plural Noun"

        # General -es rule
        return base, "Plural Noun"

    # Regular plural ending in -s
    # Example: cars -> car
    if word.endswith("s") and len(word) > 1:
        return word[:-1], "Plural Noun"

    # Singular noun
    return word, "Singular"


# -----------------------------------------
# Test words
# -----------------------------------------

words = [
    "car",
    "cars",
    "box",
    "boxes",
    "city",
    "cities",
    "child",
    "children",
    "man",
    "men",
    "woman",
    "women",
    "mouse",
    "mice",
    "book",
    "books"
]

print("Morphological Analysis")
print("-" * 40)

for word in words:
    root, category = parser(word)
    print(f"{word:10} -> {root:10} -> {category}")


# -----------------------------------------
# WordNet Verification
# -----------------------------------------

print("\nWordNet Verification")
print("-" * 40)

for word in words:
    synsets = wn.synsets(word, pos=wn.NOUN)

    if synsets:
        print(f"{word:10} -> Found in WordNet")
    else:
        print(f"{word:10} -> Not found in WordNet")
```

Error Analysis of Morphological Feature Extraction

```python
# Question 7: Error Analysis of Morphological Feature Extraction

import time
import nltk

from nltk.stem import PorterStemmer
from sklearn.datasets import fetch_20newsgroups
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Download required NLTK data
nltk.download("punkt")


# ------------------------------------------------
# 1. Load 20 Newsgroups Dataset
# ------------------------------------------------

categories = [
    "sci.space",
    "rec.sport.baseball",
    "comp.graphics",
    "talk.politics.misc"
]

train_data = fetch_20newsgroups(
    subset="train",
    categories=categories,
    remove=("headers", "footers", "quotes")
)

test_data = fetch_20newsgroups(
    subset="test",
    categories=categories,
    remove=("headers", "footers", "quotes")
)

print("Number of training documents:", len(train_data.data))
print("Number of testing documents:", len(test_data.data))


# ------------------------------------------------
# 2. Original Pipeline
# ------------------------------------------------

print("\n--- ORIGINAL PIPELINE ---")

start_time = time.time()

vectorizer_original = CountVectorizer(
    stop_words="english"
)

X_train_original = vectorizer_original.fit_transform(
    train_data.data
)

X_test_original = vectorizer_original.transform(
    test_data.data
)

original_vocab = vectorizer_original.get_feature_names_out()

model_original = LogisticRegression(
    max_iter=1000
)

model_original.fit(
    X_train_original,
    train_data.target
)

pred_original = model_original.predict(
    X_test_original
)

original_accuracy = accuracy_score(
    test_data.target,
    pred_original
)

original_time = time.time() - start_time

print("Vocabulary size before correction:",
      len(original_vocab))

print("Classification accuracy before correction:",
      original_accuracy)

print("Processing time before correction:",
      round(original_time, 2), "seconds")


# ------------------------------------------------
# 3. Stemming Function
# ------------------------------------------------

stemmer = PorterStemmer()


def stem_text(text):
    words = text.lower().split()

    stemmed_words = [
        stemmer.stem(word)
        for word in words
    ]

    return " ".join(stemmed_words)


# ------------------------------------------------
# 4. Apply Stemming BEFORE Feature Extraction
# ------------------------------------------------

print("\n--- CORRECTED PIPELINE ---")

start_time = time.time()

stemmed_train = [
    stem_text(text)
    for text in train_data.data
]

stemmed_test = [
    stem_text(text)
    for text in test_data.data
]


# ------------------------------------------------
# 5. Feature Extraction
# ------------------------------------------------

vectorizer_corrected = CountVectorizer(
    stop_words="english"
)

X_train_corrected = vectorizer_corrected.fit_transform(
    stemmed_train
)

X_test_corrected = vectorizer_corrected.transform(
    stemmed_test
)

corrected_vocab = (
    vectorizer_corrected.get_feature_names_out()
)


# ------------------------------------------------
# 6. Train Classifier
# ------------------------------------------------

model_corrected = LogisticRegression(
    max_iter=1000
)

model_corrected.fit(
    X_train_corrected,
    train_data.target
)


# ------------------------------------------------
# 7. Prediction
# ------------------------------------------------

pred_corrected = model_corrected.predict(
    X_test_corrected
)


corrected_accuracy = accuracy_score(
    test_data.target,
    pred_corrected
)

corrected_time = time.time() - start_time


# ------------------------------------------------
# 8. Results
# ------------------------------------------------

print("Vocabulary size after correction:",
      len(corrected_vocab))

print("Classification accuracy after correction:",
      corrected_accuracy)

print("Processing time after correction:",
      round(corrected_time, 2), "seconds")


# ------------------------------------------------
# 9. Comparison
# ------------------------------------------------

print("\n========== COMPARISON ==========")

print(
    f"Vocabulary before correction : "
    f"{len(original_vocab)}"
)

print(
    f"Vocabulary after correction  : "
    f"{len(corrected_vocab)}"
)

print(
    f"Accuracy before correction   : "
    f"{original_accuracy:.4f}"
)

print(
    f"Accuracy after correction    : "
    f"{corrected_accuracy:.4f}"
)

print(
    f"Time before correction       : "
    f"{original_time:.2f} seconds"
)

print(
    f"Time after correction        : "
    f"{corrected_time:.2f} seconds"
)
```
